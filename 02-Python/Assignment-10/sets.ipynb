```python
# Q21. Unique Values

numbers = [10, 20, 10, 30, 40, 20, 50]

unique_numbers = set(numbers)
unique_count = len(unique_numbers)

print("Unique Values:", unique_numbers)
print("Count:", unique_count)
```

```python
# Q22. Set Operations

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

union = a | b
intersection = a & b
difference = a - b
symmetric_difference = a ^ b

print("Union:", union)
print("Intersection:", intersection)
print("A - B:", difference)
print("Symmetric Difference:", symmetric_difference)
```

```python
# Q24. Common Students

python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}

both = python_students & sql_students
exactly_one = python_students ^ sql_students

print("Both Courses:", both)
print("Exactly One Course:", exactly_one)
```

```python
# Q25. Subset and Superset

required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}

is_subset = required_skills.issubset(candidate_skills)
is_superset = candidate_skills.issuperset(required_skills)

print("Subset:", is_subset)
print("Superset:", is_superset)
```

```python
# Q26. Missing Numbers

numbers = {1, 2, 4, 6, 8, 10}

all_numbers = set(range(1, 11))
missing = all_numbers - numbers

print("Missing Numbers:", missing)
```

```python
# Q27. Three-Set Analysis

python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python & sql & aws

python_sql_only = (python & sql) - aws

at_least_two = (python & sql) | (python & aws) | (sql & aws)

print("All Three:", all_three)
print("Python and SQL Only:", python_sql_only)
print("At Least Two:", at_least_two)
```

```python
# Q28. Transaction Analysis

day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1 & day2 & day3

exactly_one_day = (day1 ^ day2 ^ day3) - all_days

at_least_two_days = (day1 & day2) | (day1 & day3) | (day2 & day3)

print("All Days:", all_days)
print("Exactly One Day:", exactly_one_day)
print("At Least Two Days:", at_least_two_days)
```

```python
# Q29. Access Control

required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing_permissions = required_permissions - user_permissions
extra_permissions = user_permissions - required_permissions

fully_authorized = required_permissions.issubset(user_permissions)

print("Missing:", missing_permissions)
print("Extra:", extra_permissions)
print("Fully Authorized:", fully_authorized)
```

```python
# Q30. Customer Segmentation

website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store

exactly_two = (
    ((website & mobile) - store) |
    ((website & store) - mobile) |
    ((mobile & store) - website)
)

exactly_one = (
    website ^ mobile ^ store
) - exactly_two - all_three

print("Exactly One Platform:", exactly_one)
print("Exactly Two Platforms:", exactly_two)
print("All Three Platforms:", all_three)
```
