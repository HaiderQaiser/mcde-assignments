```python
# Q41. Student Performance System

students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

ali_avg = sum(students[0]["marks"]) / len(students[0]["marks"])
sara_avg = sum(students[1]["marks"]) / len(students[1]["marks"])
ahmed_avg = sum(students[2]["marks"]) / len(students[2]["marks"])

averages = (ali_avg, sara_avg, ahmed_avg)

top_average = max(averages)

top_student = (
    students[0]["name"] if top_average == ali_avg
    else students[1]["name"] if top_average == sara_avg
    else students[2]["name"]
)

names = (
    students[0]["name"],
    students[1]["name"],
    students[2]["name"]
)

unique_marks = (
    set(students[0]["marks"]) |
    set(students[1]["marks"]) |
    set(students[2]["marks"])
)

print("Averages:", averages)
print("Top Student:", top_student)
print("Names:", names)
print("Unique Marks:", unique_marks)
```

```python
# Q42. Employee Skills

employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

common_skills = (
    employees["Ali"] &
    employees["Sara"] &
    employees["Ahmed"]
)

ali_only = (
    employees["Ali"] -
    (employees["Sara"] | employees["Ahmed"])
)

all_team_skills = (
    employees["Ali"] |
    employees["Sara"] |
    employees["Ahmed"]
)

qualified = set()

if {"Python", "SQL"}.issubset(employees["Ali"]):
    qualified.add("Ali")

if {"Python", "SQL"}.issubset(employees["Sara"]):
    qualified.add("Sara")

if {"Python", "SQL"}.issubset(employees["Ahmed"]):
    qualified.add("Ahmed")

print("Common Skills:", common_skills)
print("Ali Only:", ali_only)
print("All Skills:", all_team_skills)
print("Qualified:", qualified)
```

```python
# Q43. Sales Data

sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

laptop_revenue = sales[0][1] * sales[0][2]
mouse_revenue = sales[1][1] * sales[1][2]
keyboard_revenue = sales[2][1] * sales[2][2]

revenue = laptop_revenue + mouse_revenue + keyboard_revenue

top_revenue = max(
    laptop_revenue,
    mouse_revenue,
    keyboard_revenue
)

top_product = (
    sales[0][0] if top_revenue == laptop_revenue
    else sales[1][0] if top_revenue == mouse_revenue
    else sales[2][0]
)

revenue_dict = {
    sales[0][0]: laptop_revenue,
    sales[1][0]: mouse_revenue,
    sales[2][0]: keyboard_revenue
}

prices = {
    sales[0][1],
    sales[1][1],
    sales[2][1]
}

print("Revenue:", revenue)
print("Top Product:", top_product)
print("Revenue Dict:", revenue_dict)
print("Prices:", prices)
```

```python
# Q44. Customer Orders

customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

ali_total = sum(customers["C101"]["orders"])
sara_total = sum(customers["C102"]["orders"])
ahmed_total = sum(customers["C103"]["orders"])

spending = {
    customers["C101"]["name"]: ali_total,
    customers["C102"]["name"]: sara_total,
    customers["C103"]["name"]: ahmed_total
}

highest_total = max(ali_total, sara_total, ahmed_total)

top_customer = (
    customers["C101"]["name"] if highest_total == ali_total
    else customers["C102"]["name"] if highest_total == sara_total
    else customers["C103"]["name"]
)

totals = (
    ali_total,
    sara_total,
    ahmed_total
)

print("Spending:", spending)
print("Top Customer:", top_customer)
print("Totals:", totals)
```

```python
# Q45. Course Enrollment

courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python_set = courses["Python"]
sql_set = courses["SQL"]
aws_set = courses["AWS"]

all_three = python_set & sql_set & aws_set

exactly_two = (
    ((python_set & sql_set) - aws_set) |
    ((python_set & aws_set) - sql_set) |
    ((sql_set & aws_set) - python_set)
)

exactly_one = (
    (python_set | sql_set | aws_set)
    - exactly_two
    - all_three
)

unique_students = len(
    python_set |
    sql_set |
    aws_set
)

print("Exactly Two:", exactly_two)
print("All Three:", all_three)
print("Exactly One:", exactly_one)
print("Unique Students:", unique_students)
```
