Ques 09

```python
matrix = [
    [10,20,30],
    [40,50,60],
    [70,80,90]
]
main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]
print(f"Main diagonal elements are: {main_diagonal}")
print(f"Sum of Main diagonal is: {sum(main_diagonal)}\n")

sec_diagonal = [matrix[0][2], matrix[1][1], matrix[2][0]]
print(f"Secondary diagonal elements are: {sec_diagonal}")
print(f"Sum of Main diagonal is: {sum(sec_diagonal)}\n")

middle_column = [matrix[0][1], matrix[1][1], matrix[2][1]]
print(f"Middle Column elements are: {middle_column}")
print(f"Sum of Middle Column elements is: {sum(middle_column)}")
```
