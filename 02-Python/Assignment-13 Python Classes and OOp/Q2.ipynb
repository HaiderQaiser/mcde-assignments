Q2. Secure Bank Account | Intermediate<br>
Create a BankAccount class with a public account holder name, private __balance, and private __pin. Implement
deposit(amount), withdraw(amount, pin), and get_balance(pin). Reject zero or negative deposits, withdrawals
exceeding the available balance, and incorrect PINs. Demonstrate a successful deposit, a successful withdrawal,
and two rejected operations. Extension: Explain why double underscores do not provide true security.

```python
class BankAccount:
    def __init__(self, name, balance, pin):
        self.name = name
        self.__balance = balance
        self.__pin = pin
    def deposit(self,amount):
        """
        Use this method to deposit amount.
        This method takes one argument 'amount' --> (integer).
        """
        if amount <= 0:
            print("Invalid Amount! Try again")
        else:
            self.__balance += amount
            print(f"Successfully deposited {amount} pkr.") 
    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print("Invalid Password! Try again.")
        elif pin == self.__pin:
            if amount <= 0 or amount >= self.__balance:
                print("Invalid amount to withdraw!")
            else:
                self.__balance -= amount
                print(f"Successfully withdraw {amount} pkr.")
    def get_balance(self, pin):
        """ 
        Exactly takes one argument 'pin' --> (string)
        """
        if pin == "":
            print("Password can't be empty.")
        elif pin == self.__pin:
            print(f"Your Current Balance is: {self.__balance} pkr")
        else:
            print("Invalid Password! Try again.")
# Object creation
user1 = BankAccount("Haider", 5000, "123")
user1.get_balance("123")
user1.deposit(1000)
user1.get_balance("123")
user1.withdraw(1, '1234')
```

Answer:<br

Double underscore (__) attributes in Python are not truly private. Python performs name mangling, which changes __balance to _BankAccount__balance. This prevents accidental access from outside the class, but anyone who knows the mangled name can still access it. Therefore, double underscores provide encapsulation and protection from mistakes, not real security.
