Q1. Student Result Manager | Beginner <br>
Create a Student class with a public name, protected _marks, and private __grade attributes. Initialize the
student through a constructor. Implement set_marks(marks), which accepts only values from 0 to 100, along with
get_marks() and get_grade(). Assign grade A for 80 to 100, B for 60 to 79, C for 40 to 59, and F below 40.
Demonstrate both a valid and an invalid marks update. 

```python
class Student:
    def __init__(self, name, marks, grade):
        self.name  = name
        self._marks = marks
        self.__grade  = grade
    def set_marks(self, mark):
        if mark < 0 and mark > 100:
            print("Invalid Marks! Please try again.")
        else:
            self._marks = mark
    def get_marks(self):
        print(f"Marks: {self._marks}")
    def get_grade(self):
        if self._marks >= 80:
            print(f"Grade: A") 
        elif self._marks >= 60:
            print(f"Grade: B")
        elif self._marks >= 40:
            print(f"Grade: C")
        else:
            print("You are Fail!")
```

```python
o = Student("ali", 25, "A")
```

```python
o.get_marks()
o.get_grade()
```

```python
o.set_marks(89)
o.get_marks()
o.get_grade()
```
