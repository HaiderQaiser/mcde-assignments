Q4. Vehicle Rental System | Intermediate <br>
Create a base class Vehicle with brand, model, and daily rental rate. Implement calculate_rent(days). Derive Car,
which adds a fixed insurance fee of Rs. 1,000 per rental, and Bike, which applies a 10% discount to the base
rental amount. Override calculate_rent() and reuse the parent implementation with super(). Reject rental
durations of zero or fewer days. Test both vehicle types for a three-day rental.

```python
class Vehicle:
    def __init__(self, brand, model, rentalrate):
        self.brand = brand
        self.model = model
        self.rentalrate = rentalrate

    def calculate_rent(self, days):
        if days <= 0:
            raise ValueError("Rental duration must be greater than zero.")
        return self.rentalrate * days


class Car(Vehicle):
    def calculate_rent(self, days):
        return super().calculate_rent(days) + 1000


class Bike(Vehicle):
    def calculate_rent(self, days):
        return super().calculate_rent(days) * 0.90


car = Car("Toyota", "Corolla", 3000)
bike = Bike("Honda", "Shine", 500)

print(f"Car rent for 3 days: Rs. {car.calculate_rent(3):.2f}")
print(f"Bike rent for 3 days: Rs. {bike.calculate_rent(3):.2f}")
```
